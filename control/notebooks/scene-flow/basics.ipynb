# Scene Flow 101

The goal of this notebook is to understand the basics of scene flow & understand how Isaac Lab might be useful in building out piples to evaluate scene-flow based VLA methods. 

The most basic primitive of scene flow is the process of extracting points from a rendered scene and computing their ground-truth trajectories. This can be referred to as the "GT Flow". The task of a scene flow model is to predict how these points evolve over time.
